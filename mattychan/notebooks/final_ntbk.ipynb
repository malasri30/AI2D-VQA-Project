```python
import os
import urllib.request
 
GITHUB_USER   = "adityamatale"      # e.g. "mattychan"
GITHUB_REPO   = "AI2D-VQA-Project"     # e.g. "AI2D-VQA-Project"
GITHUB_BRANCH = "main"               # or "master", or whatever branch has the notebooks
NOTEBOOKS_DIR_IN_REPO = "mattychan/notebooks"  # folder inside the repo where the .ipynb files live (leave "" if they're at repo root)
 
RAW_BASE = f"https://raw.githubusercontent.com/{GITHUB_USER}/{GITHUB_REPO}/{GITHUB_BRANCH}"
if NOTEBOOKS_DIR_IN_REPO:
    RAW_BASE = f"{RAW_BASE}/{NOTEBOOKS_DIR_IN_REPO}"
 
print(f"✓ Will fetch from: {RAW_BASE}")
```

```python
NOTEBOOK_FILES = [
    "00_environment_setup.ipynb"
    "01_eda_and_data_analysis.ipynb"
    "02_data_preprocessing.ipynb"
    "03_phase1_zeroshot.ipynb"
    "04_tuningA_qwen2vl_unsloth_kaggle.ipynb"
    "04_tuningA_qwen2vl.ipynb"
    "05_tuningB_qwen2vl_unsloth_kaggle.ipynb"
    "06_comparison_and_analysis.ipynb"
    "07_eval_metrics.ipynb"
    "08_attention_maps.ipynb"
    "09_inference.ipynb"
]
 
print(f"{len(NOTEBOOK_FILES)} notebooks queued for download:")
for f in NOTEBOOK_FILES:
    print(f"  {f}")
```

```python
def download_file(url, out_path):
    try:
        urllib.request.urlretrieve(url, out_path)
        size_kb = os.path.getsize(out_path) / 1024
        print(f"  ✓ {out_path}  ({size_kb:.1f} KB)")
        return True
    except Exception as e:
        print(f"  ✗ {out_path} failed: {e}")
        return False
 
print("✓ Helper ready")
```

```python
OUTPUT_DIR = "downloaded_notebooks"
os.makedirs(OUTPUT_DIR, exist_ok=True)

results = {}
for fname in NOTEBOOK_FILES:
    url = f"{RAW_BASE}/{fname}"
    out_path = os.path.join(OUTPUT_DIR, fname)
    results[fname] = download_file(url, out_path)
 
n_ok = sum(results.values())
print(f"\n{n_ok}/{len(NOTEBOOK_FILES)} downloaded successfully")
 
failed = [f for f, ok in results.items() if not ok]
if failed:
    print("\n⚠ Failed downloads — check GITHUB_USER/REPO/BRANCH/NOTEBOOKS_DIR_IN_REPO in cell 01,")
    print("  or confirm these filenames/paths exist in the repo:")
    for f in failed:
        print(f"  {f}")
```

```python
print("Notebooks present in current directory:")
for fname in NOTEBOOK_FILES:
    status = "✓" if os.path.exists(fname) else "✗ MISSING"
    print(f"  {status}  {fname}")
```

```python
EXTRA_FILES = [
    "metadata_generator.py",
]
 
for fname in EXTRA_FILES:
    url = f"{RAW_BASE}/{fname}"
    download_file(url, fname)
 
# %%
print("\n✓ Done. All notebooks/scripts should now be sitting next to this file — open them individually to run.")
```
